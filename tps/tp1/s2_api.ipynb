## Utiliser europeana pour trouver des images de personnes mentionnées dans Camille

https://europeana.atlassian.net/wiki/spaces/EF/pages/2385313793/Europeana+APIs+Documentation

```python
api_key = '' # les terms of use interdisent de partager la clé https://www.europeana.eu/en/rights/terms-of-use
mot_recherche = 'princesse Clotilde' # elle est mentionnée dans le fichier KB_JB92_1873-02-06_01-00002

base_url = 'https://api.europeana.eu/record/v2/search.json?'

parametres = {
    'query': mot_recherche, # cherche le mot clé au début du code
    'media': 'true', # ne montre que des objets qui possèdent un lien
    'qf': 'TYPE:IMAGE', # introduit une facette pour avoir tous les objets de type image
    'wskey': api_key # utilise la clé donnée au début du code
}

resp = requests.get(base_url, parametres)

try:
    data = resp.json()
except resp.status_code != 200: # sortir s'il n'y a pas de succès de la requête
    print(f"Erreur: {resp.status_code}")
for item in data.get('items', []):
    print(f"Title: {item.get('title', ['N/A'])[0]}") # le titre de l'oeuvre (toute la ligne), pris de la première entrée
    print(f"Creator: {item.get('dcCreator', ['N/A'])[0]}") # le nom de l'auteur
    print(f"id: {item.get('id')}") # dans le format: dataset_ID/local_ID. Les deux forment l'unique identifier (Europeana ID)
    print(f"Thumbnail: {item.get('edmPreview')[0]}") # miniature de l'image
    print(f"Link: {item.get('link')}") # lien vers les métadonnées de l'objet
    print('-' * 50) # introduit une ligne tiretée entre les objets (simplifie la lecture des résultats)
    
```
