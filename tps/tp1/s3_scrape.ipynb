# Scraping: récupération des 51 fichiers Camille



## Imports

```python
import os
import re
import time
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Téléchargement de tous les fichiers



```python
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/39.0.2171.95 Safari/537.36'}

fichiers = [] # renommé pour plus de cohérence avec les consignes du TP

root_url = f"https://max.de.wilde.web.ulb.be/camille/" # URL du TP1
response = requests.get(root_url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
for link in soup.find_all("a"): # Il ne s'agit pas de la classe ici
    title = link.text.strip()
    url = root_url + link.get('href') 
    fichiers.append([url, title])
    
```

```python
# Affichage du nombre de fichiers récupérés pour vérifier
len(fichiers)
```

```python
# Création d'un dossier "Camille" dans tp1 pour y stocker les pdf (s'il n'existe pas déjà)
cam_path = '../tp1/Camille'
if not os.path.exists(cam_path):
    os.mkdir(cam_path)
```

```python
# Création d'un data frame pour y stocker les liens et les noms des fichiers
df = pd.DataFrame(fichiers, columns=['link', 'title'])
# Afficher le data frame -> montre bien 51 lignes
df
```

```python
# Récupérer les fichiers listés dans le data frame
for url in df['link']:
    filename = url.split("/")[-1]
# Récupérer le fichier dans la RAM
    response = requests.get(url)
# Ajouter dans le "for" pour que tous les fichiers soient écrits sur le disque
    with open(os.path.join(cam_path, filename), 'wb') as f:
        f.write(response.content)
```
